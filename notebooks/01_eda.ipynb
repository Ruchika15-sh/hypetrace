```python
import pandas as pd
import pandas.plotting
pandas.plotting.register_matplotlib_converters()  
import matplotlib.transforms as transforms

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python


trends = pd.read_csv("../data/raw/google_trends_brands.csv")
stocks = pd.read_csv("../data/raw/stock_prices.csv")
events = pd.read_excel("../data/raw/event.xlsx")

trends.head()
```

```python
print(trends.shape, stocks.shape, events.shape)
print(trends.dtypes)
print(events.dtypes)
```

```python
print(trends.isnull().sum())
print(stocks.isnull().sum())
print(events.isnull().sum())
```

```python
#chnging dataype from str to date
trends['date'] = pd.to_datetime(trends['date'])
stocks['date'] = pd.to_datetime(stocks['date'])

print(trends.dtypes)
print(stocks.dtypes)
```

```python
# interest over time for each brand

plt.figure(figsize=(14,6))

for brand in trends['brand'].unique():
    subset = trends[trends['brand'] == brand]
    plt.plot(subset['date'], subset['search_interest'], label=brand)

plt.legend()
plt.title("Search Interest Over Time by Brand")
plt.xlabel("Date")
plt.ylabel("Search Interest (0-100)")

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# stocks over time
plt.figure(figsize=(14,6))

for brand in stocks['brand'].unique():
    subset = stocks[stocks['brand'] == brand]
    plt.plot(subset['date'], subset['close_price'], label=brand)

plt.legend()
plt.title("Stock Price Over Time By Brand")    
plt.xlabel("Date")
plt.ylabel("Close Price")

ax = plt.gca()
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
